```python
import requests
import json
from pyspark.sql import functions as F
from pyspark.sql.types import StructType, StructField, StringType, IntegerType, TimestampType
from datetime import datetime
```

```python
# CONFIG

BASE_URL = "https://api.openf1.org/v1/drivers"
SEASON = 2025
BRONZE_PATH = "db_course_ls.bronze.openf1_drivers"
TABLE_NAME = "bronze.openf1_drivers"
```

```python
# API CALL

response = requests.get(BASE_URL)
response.raise_for_status()

data = response.json()

print(f"Records received: {len(data)}")
```

```python
# Convert to Spark DataFrame
df = spark.createDataFrame(data)

# Add ingestion metadata

df = (
    df
    .withColumn("season", F.lit(SEASON))
    .withColumn("ingest_timestamp", F.current_timestamp())
)



display(df)
```

```python
# Write Data to Delta Bronze
(
    df.write
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable(BRONZE_PATH)
)
```
